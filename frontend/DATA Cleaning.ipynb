## Data preprocessing

```python

import pandas as pd
import numpy as np

# 1. Load dataset
df = pd.read_csv("data/delhi_pilot.csv")

print("Original Dataset:")
print(df.head())

# 2. Convert pollutant values to numeric
pollutants = ["pm25", "pm10", "no2"]

for col in pollutants:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 3. Check missing values
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# 4. Remove duplicate station-date records
df = df.drop_duplicates(subset=["date", "city", "station"])

# 5. Replace negative pollution values with NaN
for col in pollutants:
    df.loc[df[col] < 0, col] = np.nan

# 6. Handle missing values
# For a small academic demonstration:
# Remove rows missing any required pollutant.
df = df.dropna(subset=pollutants)

# 7. Convert date column to datetime
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df = df.dropna(subset=["date"])

# 8. Sort data
df = df.sort_values(["station", "date"])

# 9. Validate
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nFinal Dataset Shape:")
print(df.shape)

# 10. Save clean data
df.to_csv("data/delhi_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
